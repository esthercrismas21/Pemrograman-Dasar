Lima latihan kecil (data kecil)

Pakai skor = [80, 65, 90, 75, 70, 85]. Tulis kode yang (1) mencetak skor pertama dan skor terakhir (pakai indeks negatif untuk yang terakhir), (2) mencetak rata-rata, (3) menghitung berapa skor di atas rata-rata, (4) mencetak posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1), dan (5) membuat list baru dengan setiap skor ditambah 5.

```python
# Latihan 1: Lima Latihan Kecil (Data Kecil)

# Membuat list skor mahasiswa
skor = [80, 65, 90, 75, 70, 85]

# (1) Menampilkan skor pertama dan terakhir
print("Skor pertama:", skor[0])           # Skor Pertama
print("Skor terakhir:", skor[-1])         # Skor Terakhir

# (2) Menghitung rata-rata skor
rata_rata = sum(skor) / len(skor)         # Mencari rata-rata
print("Rata-rata skor:", rata_rata)       # Menampilkan rata-rata

# (3) Menghitung jumlah skor di atas rata-rata
jumlah_di_atas = 0

for nilai in skor:                        # Perulangan
    if nilai > rata_rata:
        jumlah_di_atas += 1

print("Jumlah skor di atas rata-rata:", jumlah_di_atas)

# (4) Mencari posisi skor terendah
skor_terendah = min(skor)
nomor_mahasiswa = skor.index(skor_terendah) + 1

print("Skor terendah:", skor_terendah)
print("Mahasiswa dengan skor terendah adalah mahasiswa ke-", nomor_mahasiswa)

# (5) Membuat list baru dengan setiap skor ditambah 5
skor_baru = []

for nilai in skor:
    skor_baru.append(nilai + 5)

print("Skor setelah ditambah 5:", skor_baru)
```

Laporan produksi bulanan (portofolio)

Dengan list energi 30 hari (salin dari blok kode di atas), tulis program yang mencetak: total dan rata-rata energi; tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya); jumlah hari di bawah target yang Anda masukkan; setiap hari yang rata-rata bergerak 7 harinya turun di bawah 97% rata-rata bulanan; serta total energi per minggu (minggu 1–4, ditambah sisa hari).

```python
# Latihan 2: Laporan Produksi Bulanan (Portofolio)

# Data produksi energi selama 30 hari (MWh)
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

# 1. Menghitung total dan rata-rata energi
total_energi = sum(energi)
rata_rata = total_energi / len(energi)

print("Total energi:", total_energi, "MWh")               # Mencetak hasil
print("Rata-rata energi:", round(rata_rata, 2), "MWh")

# 2. Mencari tiga hari dengan energi tertinggi dan terendah
data_hari = list(enumerate(energi, start=1))

tertinggi = sorted(data_hari, key=lambda x: x[1], reverse=True)[:3]
terendah = sorted(data_hari, key=lambda x: x[1])[:3]

print("\nTiga hari dengan energi tertinggi:")             # Mencetak hasil
for hari, nilai in tertinggi:                             # Perulangan
    print("Hari", hari, ":", nilai, "MWh")

print("\nTiga hari dengan energi terendah:")              # Mencetak hasil
for hari, nilai in terendah:                              # Perulangan
    print("Hari", hari, ":", nilai, "MWh")

# 3. Menghitung jumlah hari di bawah target
target = float(input("\nMasukkan target energi harian (MWh): "))  # Masukkan dari pengguna

hari_di_bawah = []
for i in range(len(energi)):                              # Perulangan
    if energi[i] < target:
        hari_di_bawah.append(i + 1)

print("Jumlah hari di bawah target:", len(hari_di_bawah))   # Mencetak hasi
print("Hari yang di bawah target:", hari_di_bawah)

# 4. Memeriksa rata-rata bergerak 7 hari
batas = rata_rata * 0.97

print("\nHari dengan rata-rata bergerak 7 hari di bawah 97% rata-rata bulanan:")    # Mencetak hasil
for i in range(6, len(energi)):                               # Perulangan
    rata_7 = sum(energi[i - 6 : i + 1]) / 7

    if rata_7 < batas:                                        # Percabangan
        print("Hari", i + 1, ":", round(rata_7, 2), "MWh")

# 5. Menghitung total energi per minggu
print("\nTotal energi per minggu:")                          # Mencetak keterangan

for minggu in range(4):                                      # Perulangan
    awal = minggu * 7
    akhir = awal + 7
    total_minggu = sum(energi[awal:akhir])
    print("Minggu", minggu + 1, ":", total_minggu, "MWh")

# Menghitung sisa hari (hari ke-29 dan 30)
sisa_hari = sum(energi[28:30])
print("Sisa hari (29-30):", sisa_hari, "MWh")                 # Mencetak hasil
```
