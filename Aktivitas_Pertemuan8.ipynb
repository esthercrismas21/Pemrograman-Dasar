Yang akan ditampilkan dari empat baris terakhir print dari tebakan saya adalah:

2. Untuk suhu dua terakhir, yang muncul adalah angka 33.

**Hal Yang Membuat Terkejut**

Hal yang membuat saya terkejut adalah cara pengambilan nilai dari belakang menggunakan index negatif. Ternyata, kita tidak hanya bisa mengakses angka dari depan saja tetapi bisa juga dari belakang. Aktivitas 6C

Cari bug-nya

harga = [12, 15, 9] punya 3 elemen. Seorang mahasiswa ingin harga terakhir dan menulis harga[len(harga)]. Tebak apa yang terjadi, lalu jalankan di dalam try/except (ubah blok di atas bila perlu) dan tuliskan baris yang benar.

Yang akan terjadi adalah IndexError, sesuai namanya ini disebabkan oleh index yang salah. Kesalahan ada pada harga[len(harga)], karena jumlah elemen dalam list hanya 3 dan index dimulai dari 0, sehingga index terakhir adalah 2 padahal len menghitung jumlah elemen sehingga program akan mencoba mengakses index 3. Untuk memperbaikinya, gunakan harga[len(harga)-1] agar program mengambil elemen terakhir dengan benar. Sehingga, hasil akhir program yang benar adalah 9.

```python
# Aktivitas 6C - Cari Bug-nya

harga = [12, 15, 9]

print("Indeks yang sah: 0 sampai", len(harga) - 1)

# Dalam try expectt
try:
    print("Harga:", harga[len(harga)])  # Indeks tidak tersedia
except IndexError as error:
    print("IndexError:", error)

# Cara aman/benar mengambil elemen terakhir
print("Harga terakhir yang benar:", harga[len(harga) - 1])
```

Aktivitas 6E

Salinan atau bukan?

Tanpa menjalankan kode, tebak isi x, y, dan z setelah baris-baris ini, lalu periksa dengan blok di atas: x = [10, 20, 30], y = x, z = x[:], y.append(40), z.append(50), x.insert(0, 5). Mengapa z berbeda?

Hasil yang akan muncul pada layar adalah:
1. Isi x akan sesuai dengan dalam kurung yaitu 10, 20 dan 30.
2. Isi y akan menggunakan list yang sama dengan variabel x karena y = x sehingga ketika ada perubahan x maka y juga akan berubah, begitu juga sebaliknya.
3. Untuk isi z akan memuat salinan dari list x, namun z akan memiliki list sendiri sehingga kalau ada perubahan pada z, tidak akan ada perubahan pada x.
4. Kode program y.append(40) akan menambahkan angka 40 ke list yang digunakan oleh x dan y. Jadi, akan berubah menjadi 10, 20, 30, 40.
5. Kode program z.append(50) akan menambahkan angka 50 ke list z saja. maka isi z menjadi 10, 20, 30, 50, tapi x dan y tidak akan berubah karena salinannya berubah.
5. Kode program x.insert(0, 5) akan menambahkan angka 5 pada indeks 0 atau posisi awal dalam list x sesuai permintaan kode program. Karena x dan y pakai list yang sama maka perubahan akan ada di y juga.
- Hasil akhir dari setiap variabel adalah: x = [5, 10, 20, 30, 40]. y = [5, 10, 20, 30, 50]. dan z = [10, 20, 30, 50].

- z berbeda karena digunakan [:] untuk menyalin seluruh isi list x menjadi sebuah list baru yang berbeda. Jadi, z memiliki list sendiri dan tidak terhubung dengan x. Sedangkan, y tidak memakai [:] sehingga x dan y merupakan satu list yang sama.

```python
# Aktivitas 6E: Salinan atau Bukan?

# Membuat list awal
x = [10, 20, 30]

# y menggunakan list yang sama dengan x
y = x

# z membuat salinan baru dari list x
z = x[:]

# Menambahkan angka 40 ke list yang digunakan oleh x dan y
y.append(40)

# Menambahkan angka 50 hanya ke list z
z.append(50)

# Menambahkan angka 5 pada indeks ke-0 list x
# Perubahan ini juga terlihat pada y karena keduanya menggunakan list yang sama
x.insert(0, 5)

# Menampilkan isi masing-masing variabel
print("Isi x:", x)
print("Isi y:", y)
print("Isi z:", z)
```

Aktivitas 6F

Tebak dulu, baru jalankan

Lanjutkan tabel secara manual untuk dua elemen terakhir (75 dan 70). Berapa total akhir dan rata-ratanya? Berapa skor yang bernilai 75 atau lebih? Lalu jalankan blok berikut.

Hasil yang akan ditampilkan oleh program adalah:
1. Total akhir setelah dilakukan penjumlahan menggunakan for adalah 380. Digunakan perulangan for untuk menjumlahkan nilai satu per satu.
2. Rata-rata dari seluruh data adalah 76 dengan menggunakan fungsi len dan operator pembagian.
3. Skor yang bernilai 75 atau lebih ada 3 dari jumlah elemen 5, digunkaan operator perbandingan untuk menemukannya.

```python
# Aktivitas 6F

skor = [80, 65, 90, 75, 70]

total = 0                    # Menyimpan jumlah seluruh skor
jumlah_lulus = 0             # Menghitung skor 75 atau lebih

for nilai in skor:
    total = total + nilai    # Menambahkan skor ke total

    if nilai >= 75:          # Memeriksa apakah skor 75 atau lebih
        jumlah_lulus = jumlah_lulus + 1

    print("total sementara:", total)

rata_rata = total / len(skor)  # Menghitung rata-rata

print("Total            :", total)
print("Rata-rata        :", rata_rata)
print("Jumlah Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

Aktivitas 6G

Ganti ukuran jendela

Ubah blok di atas menjadi rata-rata bergerak 5 hari (mulai perulangan di indeks 4, slice suhu[i - 4 : i + 1], bagi dengan 5). Berapa rata-rata yang Anda dapat dari 7 nilai? Mengapa?

Hasil yang akan didapatkan setelah ukuran jendela diubah menjadi 5 hari adalah 30.2, 30.0, 30.2. Dari 7 nilai suhu yang ada, hanya dihasilkan 3 rata-rata saja karena setiap perhitungan rata-rata perlu 5 suhu secara berurutan. Jendela perhitungan bergeser satu posisi setiap kali satu putaran. Oleh karena itu, jumlah rata-rata yang diperoleh adalag 7-5+1=3.

```python
# Aktivitas 6G
suhu = [28, 31, 30, 33, 29, 27, 32]

# Rata-rata bergerak 5 hari
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

Aktivitas 6H

Jelaskan penurunannya

Energi hari ke-17 turun lebih dari 40%. Dalam kelompok, tuliskan tiga alasan operasional yang mungkin menjelaskannya dan data apa yang akan Anda periksa untuk memastikan masing-masing. (Petunjuk: lihat jam_operasi[16] di bagian data 30 hari, dan blok hari kerja di bawah.)

Dari hasil yang dicetak program, energi pada hari ke 17 turun sebesar 41,1% dari 6.097 MWh menjadi 3.592 MWh. Tiga alasan operasional yang mungkin menyebabkan hal ini adalah:
1. Unit pembangkit mengalami gangguan atau pemeliharaan, sehingga waktu operasi berkurang dan energi yang dihasilkan juga berkurang.
- Data yang diperiksa: Catatan gangguan, jadwal pemeliharaan dan juga laporan kondisi dari turbin dan generator.
2. Ada penurunan permintaan beban listrik sehingga pembangkit mungkin mengurangi produksi energinya.
- Data yag diperiksa: catatan permintaan listrik, beban jaringan dan intruksi pengaturan daya dari operator.
3. Kendala pada pasokan bahan bakar atau peralatan pendukung sehingga proses produksi energi terhambat.
- Data yang perlu diperiksa: catatan pasokan bahan bakar, tekananatau kualitas bahan bakar dan laporan kondisi perlatan pendukung.

```python
# Data energi harian selama 30 hari (MWh)
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

hari = list(range(1, 31))
jam_operasi = [24] * 30
jam_operasi[16] = 18

NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

# Memeriksa penurunan energi harian
for i in range(1, len(energi)):
    perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100
    if perubahan < -20:
        print(f"Hari {i + 1}: {perubahan:+.1f}% dibanding hari sebelumnya")

# Memeriksa jam operasi yang kurang dari 24 jam
for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print("Hari", hari[i], ":", jam_operasi[i],
              "jam,", energi[i], "MWh")

# Membandingkan rata-rata energi hari kerja dan akhir pekan
kerja = []
libur = []

for i in range(len(energi)):
    if NAMA_HARI[i % 7] == "Sab" or NAMA_HARI[i % 7] == "Min":
        libur.append(energi[i])
    else:
        kerja.append(energi[i])

print("Rata-rata hari kerja :", round(sum(kerja) / len(kerja)), "MWh")
print("Rata-rata akhir pekan:", round(sum(libur) / len(libur)), "MWh")
```

Aktivitas 6I: Array Anda sendiri

Buat list suhu berisi 7 suhu harian (bebas). Lalu: (1) tampilkan 3 hari pertama dan 2 hari terakhir dengan slicing; (2) tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop; (3) tampilkan versi urutnya dengan sorted tanpa mengubah suhu; (4) hitung rata-rata dan cari suhu tertinggi beserta hari ke berapa; (5) buat list baru berisi suhu di atas rata-rata.

Dari program ini, saya membuat 7 data suhu harian dan mengerjakan perintah soal.
1. Saya menggunakan slicing [:3] untuk mengambil 3 data pertama dan [-2:] untuk mengambil 2 data terakhir.
2. Saya menggunakan append() untuk menambahkan suhu 33°C dan menggunakan pop() untuk mengeluarkan suhu tersebut dari data.
3. Saya menggunakan sorted untuk mengurutkan suhu dari yang terkecil tanpa mengubah list suhu asli.
4. Selanjutnya, menghitung rata-rata suhu dengan menggunakan fungsi sum (tota), operator pembagian dan fungsi len (menghitung jumlah elemen). Selain itu, digunakan juga fungsi max untuk mnecari suhu tertinggi. Lalu, saya memanggil index dengan suhu tertinggi dan menjumlahkannya dengan 1 agar sesuai dengan hari.
5. Terakhir, membuat list baru dengan suhu di atas rata-rata dengan memakai perulangan.

```python
# Aktivitas 6i

# Membuat list suhu selama 7 hari
suhu = [28, 30, 27, 32, 29, 31, 26]

# (1) Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# (2) Menambahkan satu suhu dengan append, lalu mengeluarkannya dengan pop
suhu.append(33)
print("Setelah append:", suhu)

suhu.pop()                    #Memunculkan
print("Setelah pop:", suhu)

# (3) Menampilkan suhu dalam urutan dari terkecil
print("Suhu terurut:", sorted(suhu))      #Mengurutkan
print("List suhu asli:", suhu)

# (4) Menghitung rata-rata dan mencari suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari_tertinggi = suhu.index(tertinggi) + 1

print("Rata-rata suhu:", round(rata_rata, 2))         #Mencetak hasil
print("Suhu tertinggi:", tertinggi, "°C")
print("Suhu tertinggi terjadi pada hari ke-", hari_tertinggi)

# (5) Membuat list suhu di atas rata-rata
di_atas_rata = []

for nilai in suhu:        #menggunakan perulangan for
    if nilai > rata_rata:
        di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", di_atas_rata)
```
