1. Buat program yang meminta sebuah angka, lalu tampilkan "Kelipatan 5" jika angka tersebut habis dibagi 5.


```python
#PROGRAM PEMERIKSA KELIPATAN 5
print("          PEMERIKSA KELIPATAN 5")

#Meminta pengguna memasukkan nilai
angka = int(input("Masukkan angka yang ingin diperiksa: "))

#Perhitungan
sisa = angka % 5

#Output
print("                 \nHASIL")
print(f"Angka yang diperiksa : {angka}")
print(f"Pembagi              : 5")
print(f"Sisa pembagian       : {sisa}")

#Percabangan
if sisa == 0:
    print("Status               : KELIPATAN 5")
    print(f"Perhitungan          : {angka} % 5 = 0")
    print("Kesimpulan           : Angka habis dibagi 5.")
else:
    print("Status               : BUKAN KELIPATAN 5")
    print(f"Perhitungan          : {angka} % 5 = {sisa}")
    print("Kesimpulan           : Angka tidak habis dibagi 5.")
```

Buat program untuk menentukan apakah suatu tahun merupakan tahun kabisat berdasarkan kondisi berikut:
*   Tahun yang habis dibagi 400 merupakan tahun kabisat.
*   Tahun yang habis dibagi 4 dan tidak habis dibagi 100 merupakan tahun kabisat.

```python

#PROGRAM PEMERIKSA TAHUN KABISAT
print("          PEMERIKSA TAHUN KABISAT")

#Meminta pengguna memasukkan tahun
tahun = int(input("Masukkan tahun yang ingin diperiksa: "))

#Output
print("                 \nHASIL")
print(f"Tahun yang diperiksa : {tahun}")

#Percabangan
if tahun % 400 == 0:
    print("Status               : TAHUN KABISAT")
    print("Syarat               : Habis dibagi 400.")
    print("Kesimpulan           : Bulan Februari memiliki 29 hari.")
elif tahun % 4 == 0 and tahun % 100 != 0:
    print("Status               : TAHUN KABISAT")
    print("Syarat               : Habis dibagi 4 dan tidak habis dibagi 100.")
    print("Kesimpulan           : Bulan Februari memiliki 29 hari.")
else:
    print("Status               : BUKAN TAHUN KABISAT")
    print("Syarat               : Tidak memenuhi aturan tahun kabisat.")
    print("Kesimpulan           : Bulan Februari memiliki 28 hari.")
```

```python
# Program untuk menentukan tahun kabisat
print("Program Untuk Menentukan Tahun Kabisat")

#Meminta pengguna memasukkan tahun
tahun = int(input("Masukkan tahun: "))

#Percabangan
print("\nHasil:")
if tahun % 400 == 0 or tahun % 4 == 0 and tahun % 100 != 0:
    print(tahun, "adalah tahun kabisat")
else:
    print(tahun, "bukan tahun kabisat")
```
