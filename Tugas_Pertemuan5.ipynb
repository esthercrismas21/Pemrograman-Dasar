Soal 1 : Belanja Buku

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

Total belanja, besar diskon, dan jumlah yang harus dibayar.
Uang kembalian.
Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
Apakah uang Budi cukup? (hasil True/False)
Jawaban

program python di sini
Contoh Output
Total       : 129500
Diskon      : 12950
Bayar       : 116550
Kembalian   : 33450
Per orang   : 29137
Sisa        : 2
Uang cukup? : True
Operator yang dipakai: *, //, -, %, >=.

```python
#Belanja Buku
print("Program Menghitung Belanja Buku")

#Variabel
jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

#Proses
total = jumlah_buku * harga_buku                  #Menghitung total belanja
diskon = total * diskon_persen // 100             #Menghitung Besar Diskon
bayar = total - diskon                            #Menghitung Jumlah yang harus dibayar
kembalian = uang_budi - bayar                     #Menghitung uang kembalian
per_orang = bayar // jumlah_orang                 #Menghitung jumlah iuran
sisa = bayar % jumlah_orang                       #Menghitung sisa iuaran
uang_cukup = bayar >= uang_budi                   #Menetapkan apakah uang Budi cukup

#Output
print("\n#Data")
print("Jumlah Buku     : ", jumlah_buku)
print("Harga Buku      : ", harga_buku)
print("Diskon Persen   : ", diskon_persen)
print("Uang Budi       : ", uang_budi)

print("\n#Hasil Perhitungan")
print("Total            : ", total)
print("Diskon           : ", diskon)
print("Bayar            : ", bayar)
print("Kembalian        : ", kembalian)
print("Per orang        : ", per_orang)
print("Sisa Hasil Iuran : ", sisa)
print("Uang cukup?      : ", uang_cukup)
```

Soal 2 : PLTS Atap Rumah
Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

Energi yang dihasilkan per hari (kWh).
Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
Kekurangan energi per hari terhadap kebutuhan.
Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)
Jawaban

program python di sini
Contoh Output
Energi harian (kWh) : 21.6
Energi bulanan (kWh): 648.0
Nilai listrik (Rp)  : 936165.6
Kekurangan (kWh)    : 3.4
Mencukupi?          : False
Operator yang dipakai: *, /, -, >=. Fungsi round() hanya untuk merapikan angka desimal.

```python
#Soal 2 (PLTS Atap Rumah)
print("Program PLTS Atap Rumah")

#Variabel
jumlah_panel = 12
daya_panel = 450
jam_penyinaran = 5
efisiensi_sistem = 0.8
kebutuhan_harian = 25
tarif_listrik = 1444.70

#Proses
energi_harian = jumlah_panel * daya_panel * jam_penyinaran * efisiensi_sistem / 1000  #Energi yang dihasilkan per hari (kWh)
energi_bulanan = energi_harian * 30                                                   #Energi per bulan (30 hari)
nilai_listrik = energi_bulanan * tarif_listrik                                        #Nilai rupiah listrik yang dihasilkan
kekurangan = energi_harian - kebutuhan_harian                                         #Kekurangan energi per hari terhadap kebutuhan
mencukupi = energi_harian >= kebutuhan_harian                                         #Apakah PLTS mencukupi?

#Output
print("\n#Data")
print("Jumlah Panel                 :", jumlah_panel)
print("Daya Tiap Panel              :", daya_panel)
print("Rata-rata penyinaran efektif :", jam_penyinaran)
print("Efisiensi Sistem             :", efisiensi_sistem)
print("Kebutuhan Listrik Rumah      :", kebutuhan_harian)
print("Tarif Listrik (Rp)           :", tarif_listrik)

print("\n#Hasil Perhitungan")
print("Energi Harian (kWh)    :", round(energi_harian, 2))
print("Energi Bulanan (kWh)   :", round(energi_bulanan, 2))
print("Nilai Listrik (Rp)     :", round(nilai_listrik, 2))
print("Kekurangan (kWh)       :", round(kekurangan, 2))
print("Mencukupi?             :", mencukupi)
```

Soal 3 : Generator Diesel dan Rugi Kabel
Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

Energi yang dibangkitkan per hari (kWh).
Kebutuhan solar per hari dan biayanya.
Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
Rugi daya pada kabel (P = I² × R).
Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)
Jawaban

program python di sini
Contoh Output

Solar (liter)    : 400.0
Biaya (Rp)       : 5400000.0
Tangki cukup     : 2.0 hari
Sisa solar (L)   : 200.0
Rugi kabel (W)   : 200.0
Rugi tinggi?     : True
Operator yang dipakai: *, //, %, **, >, >=, and.



```python
#Soal 3 (Generator Diesel dan Rugi Kabel)
print("Program Generator Diesel dan Rugi Kabel")

#Variabel
daya_generator      = 200
jam_operasi         = 8
kebutuhan_liter     = 0.25
harga_solar         = 13500
kapasitas_tangki    = 1000
arus_kabel          = 20
hambatan_kabel      = 0.5
arus_minimal        = 20

#Operasi
energi_harian = daya_generator * jam_operasi                                   #Energi yang dibangkitkan per hari (kWh)
kebutuhan_solar = energi_harian * kebutuhan_liter                              #Kebutuhan solar per hari
biaya_solar = kebutuhan_solar * harga_solar                                    #Biaya solar
hari_penuh = kapasitas_tangki // kebutuhan_liter                               #Berapa hari penuh tangki cukup
sisa_solar = kapasitas_tangki % kebutuhan_liter                                #Berapa liter solar yang tersisa
rugi_daya = arus_kabel ** 2 * hambatan_kabel                                   #Rugi daya pada kabel
rugi_tinggi = rugi_daya > 150 and arus_kabel >= 20                             #Menentukan kondisi

#Output
print("\n#Data")
print("Daya Generator     :", daya_generator)
print("Jam Operasi        :", jam_operasi)
print("Kebutuhan Liter    :", kebutuhan_liter)
print("Harga Solar (Rp)   :", harga_solar)
print("Kapasitas Tangki   :", kapasitas_tangki)
print("Arus Kabel         :", arus_kabel)
print("Hambatan Kabel     :", hambatan_kabel)
print("Arus Minimal       :", arus_minimal)
print("\n#Hasil Perhitungan")
print("Energi Yang Dibangkitkan Per Hari (kWh)  :", energi_harian)
print("Kebutuhan Solar Per Hari (Liter)         :", kebutuhan_solar)
print("Biaya Solar (Rp                          :", biaya_solar)
print("Jumlah Hari Penuh Tangki Cukup           :", hari_penuh, "Hari")
print("Jumlah Liter Yang Tersisa (L)            :", sisa_solar, "Liter")
print("Rugi Daya Pada Kabel (W)                 :", rugi_daya)
print("Rugi Tinggi?                             :", rugi_tinggi)
```
