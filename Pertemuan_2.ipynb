Latihan 2 — Tipe primitif & type() (mudah)
Buat satu variabel untuk masing-masing tipe: int, float, bool, str (data komponen elektronika bebas). Cetak nilai dan tipe tiap variabel dengan type().

```python
#Jawaban Latihan 2

#Tipe Data Integer
jumlah_komponen = 10
print(jumlah_komponen, type(jumlah_komponen))

#Tipe Data Float
nilai_resistor = 211.2
print(nilai_resistor, type(nilai_resistor))

#Tipe Data Bool
komponen_aktif = True
print(komponen_aktif, type(komponen_aktif))

#Tipe Data String
nama_komponen = "Resistor"
print(nama_komponen, type(nama_komponen))
```

Latihan 4 — List (sedang)

Berikut daftar pembacaan tegangan (V) sebuah panel:

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

Tugas: (a) cetak pembacaan pertama dan terakhir memakai indeks, (b) tambahkan satu pembacaan baru 222.0 dengan append(), (c) cetak jumlah total data dengan len().

```python
#Jawaban Latihan 4

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

#A. Catak Pembacaan Pertama Dan Terakhir Menggunakan Indeks
print("Pembacaan Pertama:", tegangan_harian[0])
print("Pembacaan Terakhir:", tegangan_harian[5])

#B. Menambahkan Satu Pembacaan Baru 222.0 Dengan append()
tegangan_harian.append(222.0)

#C. Cetak Jumlah Total Data Dengan len()
print("Data Setelah Ditambahkan:", tegangan_harian)
print("Jumlah Total Data:", len(tegangan_harian))
```

Latihan 5 — Dictionary (sedang)
Buat dictionary spesifikasi sebuah motor listrik (minimal: nama, daya_kW, tegangan_V, rpm). Cetak salah satu nilai lewat key-nya, lalu ubah nilai rpm (pakai =), dan cetak ulang seluruh dictionary.

```python
#Jawaban Latihan 5

motor_listrik = {
    "nama" : "Motor Induksi",
    "daya_kW" : 15,
    "tegangan_V" : 288.9,
    "rpm" : 2112,
    "tiga_fasa" : True
}
print("Nama:", motor_listrik["nama"], "Daya:", motor_listrik["daya_kW"], "Tegangan:", motor_listrik["tegangan_V"], "rpm:", motor_listrik["rpm"], "Tiga Fasa:", motor_listrik["tiga_fasa"])

#A. Mencetak Salah Satu Nilai Lewat Key
print(motor_listrik["rpm"])

#B. Mengubah Nilai rpm Memakai (=)
motor_listrik["rpm"] = 6006

#C. Mencetak Ulang Seluruh Dictionary
print("Nama:", motor_listrik["nama"], "Daya:", motor_listrik["daya_kW"], "Tegangan:", motor_listrik["tegangan_V"], "rpm:", motor_listrik["rpm"], "Tiga Fasa:", motor_listrik["tiga_fasa"])
```

Latihan 6 — Gabungan (menantang)

Kelola data tiga unit pembangkit sebagai list of dictionary, tiap unit berisi nama, kapasitas_MW, dan status. Tugas: cetak nama unit kedua, ubah status salah satu unit, tambahkan satu unit baru ke list dengan append(), lalu cetak seluruh data.


```python
# Jawaban Latihan 6

pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

#Mencetak Nama Unit Kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

#Mengubah Status Salah Satu Unit
pembangkit[0]["status"] = "pemeliharaan"
print("Status Unit Pertama:", pembangkit[0]["status"])

#Menambahkan Satu Unit Baru Ke List Dengan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 1000, "status": "operasi"})
print("Unit Yang Baru Ditambahkan:", pembangkit[3])

#Mencetak Seluruh Data
print(pembangkit)
```
