Selain itu, program ini bertujuan untuk meminta pengguna memasukkan suatu nilai yang akan dihitung nantinya dalam berbagai operasi aritmatika yang dapat dilihat perbedaan hasilnya, mulai dari operasi penjumlahan, pengurangan, perkalian, pembagian, pembagian bulat dan modulus. Dan agar memahami dan mengetahui fungsi dari masing-masing operator aritmatika serta bagaimana Python melakukan perhitungan dari berbagai fungsi operator.



```python
#Meminta Pengguna Memasukkan Dua Bilangan Bulat
bilangan1 = int(input("Masukkan Bilanga Pertama: "))
bilangan2 = int(input("Masukkan Bilangan Kedua: "))

#Hasil Operasi +, -, *, /, // dan %
print("Penjumlahan (+):", bilangan1 + bilangan2)
print("Pengurangan (-):", bilangan1 - bilangan2)
print("Perkalian (*):", bilangan1 * bilangan2)
print("Pembagian (/):", bilangan1 / bilangan2)
print("Pembagain Bulat (//):", bilangan1 // bilangan2)
print("Modulus (%):", bilangan1 % bilangan2)
```

*   Tujuan Program = Program ini bertujuan untuk memahami penggunaan operator relasional dan operator logika and dalam menentukan suatu kondisi berdasarkan lebih dari satu syarat. Program ini juga bertujuan untuk melatih penggunaan tanda kurung dalam menentukan urutan suatu ekspresi atau operator precedence.





```python
#Meminta Pengguna Memasukkan Umur dan Status Memiliki SIM (True/False)
umur = int(input("Masukkan Umur Anda: "))
status_memiliki = input("Apakah Mempunyai SIM? (True/False): ")

#Menentukan Apakah Boleh Mengemudi Sendiri?
boleh_mengemudi = (umur >= 17) and status_memiliki
print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")

#Latihan Operator Precedence
memenuhi_syarat_berkendara = ((umur >= 17) and (status_memiliki))
print(f"Memenuhi Syarat Berkendara: {memenuhi_syarat_berkendara}")
```

Program ini juga bertujuan untuk memahami penggunaan tanda kurung agar perhitungan sesuai dengan rumus yang digunakan, serta menggunakan f-string untuk menampilkan hasil perhitungan dengan dua angka di belakang koma.



```python
1
#Meminta Pengguna Memasukkan Nilai Energi Input dan Energi Output (MJ)
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

#Menghitung Efisiensi Thermal
efisiensi = (energi_output / energi_input) * 100

#Menampilkan Hasil Efisiensi Dengan f-string, Dibulatkan 2 Angka Desimal
print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```

Saat menggunakan operator or, program dapat menggabungkan kedua kondisi tersebut dan menghasilkan True jika salah satu kondisi terpenuhi, sedangkan False jika kedua kondisi tidak terpenuhi. Program ini juga bertujuan untuk memahami bagaimana operator relasional dan logika dapat digunakan bersama untuk melakukan pengecekan kondisi pada permasalahan yang berkaitan dengan sistem energi.



```python
#Meminta Pengguna Memasukkan Suhu Boiler dan Beban Operasi (%)
suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

#Menggunakan Operator Relasional dan Logika
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

#Tampilkan Hasil
print(f"Status unit: Perlu Maintenance = {perlu_maintenance}")
```
