**Langkah 1.1:** prosedur kop laporan. Prosedur cetak_kop mencetak judul di antara dua garis. Panjang garis menyesuaikan panjang judul.

```python
def cetak_kop(judul):
    """Prosedur: mencetak judul laporan di antara dua garis."""
    garis = "=" * (len(judul) + 4)
    print(garis)
    print("  " + judul)
    print(garis)


cetak_kop("LAPORAN HARIAN UNIT 1")
print("Energi 2 November: 6098 MWh")
print()
cetak_kop("RINGKASAN MINGGUAN")
```

**Langkah 1.4:** mengembalikan beberapa nilai. Fungsi ringkasan mengembalikan tiga nilai sekaligus dari array energi 7 hari. Hasilnya langsung dipecah ke tiga variabel.

```python
def ringkasan(data):
    """Kembalikan (terendah, tertinggi, rata-rata) dari sebuah array."""
    return min(data), max(data), sum(data) / len(data)


energi = [5576, 6098, 6078, 6095, 6074, 6077, 5564]     # MWh, hari 1-7

terendah, tertinggi, rata = ringkasan(energi)
print(f"Terendah  : {terendah} MWh (hari {energi.index(terendah) + 1})")
print(f"Tertinggi : {tertinggi} MWh (hari {energi.index(tertinggi) + 1})")
print("Rata-rata :", round(rata, 1), "MWh")
print("Satu nilai:", ringkasan([7920]))
```

**Langkah 2.1:** docstring dan nilai default. capacity_factor memakai default Unit 1 (330 MW) selama satu hari (24 jam). Fungsi yang sama dipakai untuk satu hari, satu minggu, dan unit lain hanya dengan mengubah argumen.

```python
def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) = energi / (daya terpasang x jam) x 100."""
    return energi_mwh / (daya_mw * jam) * 100


energi = [5576, 6098, 6078, 6095, 6074, 6077, 5564]     # MWh, hari 1-7

print(capacity_factor.__doc__)
print("Hari 2         :", round(capacity_factor(energi[1]), 1), "%")
print("Seminggu       :", round(capacity_factor(sum(energi), jam=24 * 7), 1), "%")
print("Unit 2 (600 MW):", round(capacity_factor(12150, daya_mw=600), 1), "%")
```

**Langkah 2.3:** uji assert dan argumen tertukar. Fungsi efisiensi diuji dulu dengan hitungan manual. Lalu fungsi dipanggil tiga kali: urutan benar, urutan tertukar, dan dengan kata kunci. Pemeriksaan “masuk akal” (PLTU batubara biasanya 25–45 %) menangkap argumen yang tertukar.

```python
def efisiensi_termal(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Efisiensi termal (%) dari batubara (t) dan energi terkirim (MWh)."""
    heat_rate = batubara_t * nilai_kalor_kcal_kg / energi_mwh       # kcal/kWh
    return 860 / heat_rate * 100


# Hitungan manual: 1 t x 4400 kcal/kg / 2 MWh = 2200 kcal/kWh -> 860 / 2200 = 39,09 %
assert round(efisiensi_termal(1, 2), 2) == 39.09
print("Uji lulus")

e1 = efisiensi_termal(2998, 5576)                          # urutan benar
e2 = efisiensi_termal(5576, 2998)                          # tertukar!
e3 = efisiensi_termal(energi_mwh=5576, batubara_t=2998)    # kata kunci
for e in [e1, e2, e3]:
    print(round(e, 2), "%", "masuk akal" if 25 <= e <= 45 else "TIDAK masuk akal")
```

**Praktikum 3: Modularisasi perhitungan pembangkit**

**Langkah 3.1:** empat rumus, empat fungsi, empat uji. Ketik keempat fungsi KPI dari bagian 8, lalu uji masing-masing dengan nilai hitungan manual. Program hanya mencetak satu baris jika semua uji lulus.

```python
def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh) dari batubara (t) dan energi terkirim (MWh)."""
    if energi_mwh <= 0:
        raise ValueError("energi_mwh harus positif")
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate. 1 kWh = 860 kcal."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_mw * jam) * 100


def konsumsi_spesifik(batubara_t, energi_mwh):
    """Konsumsi batubara spesifik (t/MWh)."""
    return batubara_t / energi_mwh


# Uji dengan hitungan manual
assert heat_rate(1, 2) == 2200                    # 1 x 4400 / 2
assert round(efisiensi_termal(2150), 2) == 40     # 860 / 2150 = 0,4
assert round(capacity_factor(7920)) == 100        # 330 MW x 24 jam = 7920 MWh
assert konsumsi_spesifik(100, 200) == 0.5         # 100 t / 200 MWh
print("Keempat uji lulus")

hr = heat_rate(3258, 6098)
print(f"Hari 2: HR {hr:,.0f} kcal/kWh, efisiensi {efisiensi_termal(hr):.2f} %, "
      f"CF {capacity_factor(6098):.1f} %, SCC {konsumsi_spesifik(3258, 6098):.3f} t/MWh")
```

```python
def heat_rate(batubara_t, energi_mwh, nilai_kalor_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return batubara_t * nilai_kalor_kcal_kg / energi_mwh


def efisiensi_termal(heat_rate_kcal_kwh):
    """Efisiensi termal (%) dari heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energi_mwh, daya_mw=330, jam=24):
    """Capacity factor (%) untuk suatu periode."""
    return energi_mwh / (daya_mw * jam) * 100


def status_efisiensi(efisiensi):
    """'BURUK' jika < 36,0 %, 'RENDAH' jika < 36,3 %, selain itu 'NORMAL'."""
    if efisiensi < 36.0:
        return "BURUK"
    if efisiensi < 36.3:
        return "RENDAH"
    return "NORMAL"


def cetak_baris(hari, hr, efi, cf, status):
    """Prosedur: satu baris tabel laporan."""
    print(f"{hari:>4}  {hr:>7,.0f}  {efi:>6.2f}  {cf:>5.1f}  {status}")


# --- Data Unit 1, 14-20 November 2026 ---
hari_awal = 14
energi = [5533, 5566, 6097, 3592, 6059, 6100, 6098]      # MWh
batubara = [2996, 3009, 3257, 1950, 3234, 3261, 3262]    # t

# --- Program utama ---
print(f"{'Hari':>4}  {'HR kcal':>7}  {'Efi %':>6}  {'CF %':>5}  Status")
efisiensi = []
for i in range(len(energi)):
    hr = heat_rate(batubara[i], energi[i])
    efi = efisiensi_termal(hr)
    efisiensi.append(efi)
    cetak_baris(hari_awal + i, hr, efi, capacity_factor(energi[i]), status_efisiensi(efi))

terendah = min(efisiensi)
print(f"\nRata-rata efisiensi : {sum(efisiensi) / len(efisiensi):.2f} %")
print(f"Efisiensi terendah  : {terendah:.2f} % (hari {hari_awal + efisiensi.index(terendah)})")
print(f"CF seminggu         : {capacity_factor(sum(energi), jam=24 * len(energi)):.1f} %")
```
