**SOAL**
Tahun Kabisat.
Studi Kasus.
Sebuah tahun disebut tahun kabisat jika memenuhi salah satu dari dua kondisi berikut:
a. Tahun tersebut habis dibagi 4, tetapi tidak habis dibagi 100.
b. Tahun tersebut habis dibagi 400.
Buatlah sebuah ekspresi logika kompleks untuk menentukan apakah sebuah tahun merupakan tahun kabisat.

```python
#Menentukan Tahun Kabisat
print("#Menentukan Tahun Kabisat")

#Meng-input Tahun
tahun = int(input("Masukkan Tahun:"))

#Menentukan Tahun Kabisat Memakai or dan and
tahun_kabisat = (tahun % 4 == 0 and tahun % 100 !=0) or (tahun % 400 == 0)

#Menampilkan Hasil
print("\n#Hasil Menentukan Tahun Kabisat")
if tahun_kabisat:
  print(tahun, "Merupakan Tahun Kabisat")
  print("Tahun", tahun, "memiliki 28 hari di bulan Februari.")
else:
  print(tahun, "Bukan Merupakan Tahun Kabisat")
  print("Tahun", tahun, "memiliki 29 hari di bulan.")
```

2.   Tujuan Program = Untuk membuat program yang dapat menghitung nilai praktikum mahasiswa berdasarkan beberapa variabel nilai dengan menggunakan fungsi input, operator aritmatika, serta fungsi output untuk menampilkan hasil.

**Soal** Buatlah program sederhana versi anda masing-masing dengan lingkup, fungsi input, fungsi output, dan operasi aritmatika.

```python
#Penilaian Pratikum
print("#Menghitung Nilai Pratikum Mahasiswa")

#Memasukkan Nilai (Fungsi Input)
nama_lengkap = input("Masukkan Nama Lengkap Mahasiswa:")
nilai_laporan = float(input("Masukkan Nilai Laporan:"))
nilai_keaktifan = float(input("Masukkan Nilai Keaktifan:"))
nilai_keterampilan = float(input("Masukkan Nilai Keterampilan:"))
nilai_presentasi = float(input("Masukkan Nilai Presentasi:"))
nilai_test = float(input("Masukkan Nilai Test:"))

#Proses Operasi Aritmatika (Operasi Aritmatika)
nilai_praktikum = (nilai_laporan * 0.20) + (nilai_keaktifan * 0.15) + (nilai_keterampilan * 0.25) + (nilai_presentasi * 0.15) + (nilai_test * 0.25)
nilai_rata_rata = (nilai_laporan + nilai_keaktifan + nilai_keterampilan + nilai_presentasi + nilai_test) / 5

#Menampilkan Nilai Akhir (Fungsi Output)
print("\n#Hasil Menghitung Nilai Pratikum Mahasiswa")
print("Nilai Pratikum Mahasiswa:")
print("Nama                    :", nama_lengkap)
print("Nilai Laporan           :", nilai_laporan)
print("Nilai Keaktifan         :", nilai_keaktifan)
print("Nilai Keterampilan      :", nilai_keterampilan)
print("Nilai Presentasi        :", nilai_presentasi)
print("Nilai Test              :", nilai_test)
print("Nilai Pratikum          :", nilai_praktikum)
print("Nilai Rata-Rata         :", nilai_rata_rata)

#Menentukan Kelulusan
if nilai_praktikum >= 70:
  print("\nSELAMAT", nama_lengkap, "ANDA LULUS!")
else:
  print("\nMaaf, Anda Belum Lulus,", nama_lengkap)
```
