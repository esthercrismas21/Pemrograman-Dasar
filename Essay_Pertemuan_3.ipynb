**Operator Precedence**
Operator Precedence adalah aturan yang menentukan operator mana yang akan dikerjakan terlebih dahulu ketika dalam satu ekspresi mempunyai beberapa jenis operator. Dalam situs Python, terdapat tabel yang merangkum prioritas operator, mulai dari yang tertinggi (paling mengikat) sampai prioritas terendah (paling tidak mengikat). Python menjalankan operator dengan priotitas yang tertinggi terlebih dahulu dan akan dilanjutkan dengan prioritas lebih rendah. Sedangkan, untuk operator dengan prioritas yang sama akan dijalankan dari kiri ke kanan. Hal ini penting agar perhitungan yang dilakukan oleh Python benar dan dapat digunakan. Berikut adalah urutan prioritas dalam Python untuk ekspresi aritmatika, logika, assignment, dan relasional:
1.   Tanda Kurung ()
2.   Pangkat **
1.   Perkalian *, Pembagian /, Pembagian Bulat //, Modulus %
2.   Penjumlahan +, Pengurangan -
1.   Operator Perbandingan <,>, =<, =>, ==, !=, dll.
2.   Logika NOT not
1.   Logika AND and
2.   Logika OR or










```python
#Urutan Prioritas Operator Precedence
urutan_prioritas = 10 + 5 * 2
print(urutan_prioritas)
```

Setelah dijalankan dan hasil keluar, Python menghasilkan nilai 20 dan bukan 30. Hal ini dikarenakan adanya aturan Precedence. Menurut tabel dari situs resmi Python, operator perkalian * memiliki prioritas yang lebih tinggi daripada operator penjumlahan +. Jadi Python mengerjakan perkalian terlebih dahulu walaupun operator penjumlahan berada di sebelah kiri. Jadi dapat disimpulkan, bahwa Python tidak hanya mengerjakan operasi dari kiri ke kanan. Python terlebih dahulu melihat prioritas operator.

```python
#Bagaimana Proses Python Mengeksekusi Kode

#Mengerjakan Perkalian (Prioritas Lebih Tinggi)
perkalian = 5 * 2
print("Perkalian:", perkalian)

#Mengerjakan Penjumlahan (Prioritas Lebih Rendah)
hasil = perkalian + 10
print("Hasil Akhir:", hasil)
```

**Nomor 2** Perbedaan mendasar antara operator relasional (>, <, ==, dll.) dan operator logika (and, or, not) dalam Python. Berikan satu contoh situasi nyata (boleh di luar konteks pembangkit) di mana kedua jenis operator ini harus dipakai bersamaan dalam satu ekspresi untuk menghasilkan keputusan yang benar.

**Operator Relasional** Digunakan untuk membandingkan dua nilai yang ada. Hasil dari perbandingan tersebut adalah nilai Boolean, yaitu True atau False. Operator yang digunakan antara lain lebih besar (>), lebih kecil (<), tidak sama dengan (!=), dan lainnya.
**Operator Logika** Digunakan untuk menggabungkan atau membalikkan kondisi yang nantinya akan menghasilkan nilai True atau False. Operator ini menggunakan and (kedua kondisi harus benar), or (salah satu kondisi harus benar) dan not (membalik hasil boolean).

```python
#Contoh Operator Relasional
operator_relasional = 10 > 5
print("Hasil Operator Relasional:", operator_relasional)

#Contoh Operator Logika
punya_sim = True
not punya_sim
```

```python
#Contoh Situasi Nyata

print("SISTEM AKSES LABORATORIUM")

nama = input("Masukkan nama: ")
umur = int(input("Masukkan umur: "))

terdaftar = input("Apakah kamu terdaftar praktikum? (ya/tidak): ")
kartu = input("Apakah kartu mahasiswa masih aktif? (ya/tidak): ")

terdaftar = terdaftar.lower() == "ya"
kartu = kartu.lower() == "ya"

boleh_masuk = umur >= 17 and terdaftar and kartu

print("\nHASIL PEMERIKSAAN")
print("Nama:", nama)
print("Umur:", umur)
print("Terdaftar praktikum:", terdaftar)
print("Kartu mahasiswa aktif:", kartu)
print("Boleh masuk laboratorium:", boleh_masuk)
```

**Operator** Digunakan beberapa operator relasional dan logika:

* >= adalah operator relasional untuk mengecheck apakah umur sudah 17 tahun atau lebih
*   and adalah operator logika untuk menggabungkan semua kondisi, semua kondisi harus True agar hasilnya True. Jika salah satu syarat bernilai false maka hasilnya akan False.

**Relasional & logika**
berada pada kode "boleh_masuk" dengan >= sebaagai operator relasional dan and sebagai operator logika.

**Intinya** adalah operator relasional digunakan untuk mengecheck setiap kondisi sedangkan operator logika digunakan untuk menggabungkan kondisi yang ada menjadi satu keputusan.

**Penjelasan Kasus**
Dalam forum diskusi programmer, Python.org, seorang programmer menyampaikan masalahnya saat ia menemukan hasil yang dianggapnya aneh. Ketika menggunakan kode -3 ** 2 dan mengharapkan hasil 9, Python malah mengeluarkan nilai -9. Ia mengira -3 akan dihitung duluan sebagai bilangan negatif padahal di Python operator pangkat ** punya prioritas yang lebih tinggi dari tanda negatif (-). Akibatnya, Python membaca -3 ** 1 sebagai -(3 ** 1) sehingga hasilnya -9 sedangkan yang diingkan adalah (-3)**2 yang menghasilkan 9.

**Perbaikan Kesalahan**
Perbaikan dilakukan dengan menambahkan tanda kurung pada -3 sehingga Python menganggap (-3) sebagai satu nilai yang sama yang akan dipangkatkan. Tanpa tanda kurung, Python akan menghitung pangkat terlebih dahulu daripada tanda negatif karena adanya aturan operator precedence. Oleh karena itu, penggunaan tanda kurung () akan membantu memastikan urutan perhitungan sesuai dengan yang dibutuhkan.



```python
#Perbaikan

#Kode Sebelum Perbaikan
sebelum = -3 ** 2
print("Sebelum Perbaikan:", sebelum)

#Kode Setelah Perbaikan
sesudah = (-3)**2
print("Sesudah Perbaikan:", sesudah)
```
