**Operator Aritmatika**
Mengkonversi total detik menjadi jam:Menit:Detik.
1. Total detik diterima dari keyboard
2. Melakukan konversi nilai 358200

```python
#Kode Program Dari Materi
#Konversi waktu
total_detik = 3600
jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam: {menit} Menit: {detik} Detik.")
```

```python
#Konversi Waktu dari Total Detik
print("#Konversi Waktu")

#Memasukkan nilai dari keyboard
total_detik = int(input("Masukkan Total Detik:"))

#Konversi menggunakan operator aritmatika
jam = total_detik // 3600
sisa_detik = total_detik %3600
menit = sisa_detik // 60
detik = sisa_detik % 60

#output
print("\n#Hasil Konversi Waktu")
print(f"Hasil Konversi: {jam} Jam: {menit} Menit: {detik} Detik.")
print("Jam:", jam)
print("Menit:", menit)
print("Detik:", detik)
```

**Ekspresi**
Menghitung nilai akhir mahasiswa.
1. nilai mahasiswa dan standar kelulusan harus bisa menerima input dari keyboard
2. Program dimodifikasi agar nilai mahasiswa terdiri dari 5 variabel dan semuanya diinput dari keyboard

```python
#Kode Program Dari Materi
tugas = 80
uts = 70
uas = 90
nilai_akhir = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)
print("Nilai Akhir=", nilai_akhir)

nilai_mahasiswa = 90
standar_kelulusan = 70
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
#Menghitung Nilai Akhir Mahasiswa
print("#Menghitung Nilai Akhir Mahasiswa")

#Memasukkan Nilai Mahasiswa dari Keyboard
tugas = float(input("Masukkan nilai tugas           : "))
kuis = float(input("Masukkan nilai kuis             : "))
praktikum = float(input("Masukkan nilai praktikum   : "))
uts = float(input("Masukkan nilai UTS               : "))
uas = float(input("Masukkan nilai UAS               : "))

#Proses
nilai_mahasiswa = (0.10 * tugas) + (0.10 * kuis) + (0.20 * praktikum) + (0.30 * uts) + (0.30 * uas)
standar_kelulusan = float(input("Masukkan standar kelulusan: "))
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

#Output
print("\n#HASIL PERHITUNGAN")
print(f"Nilai Tugas       : {tugas}")
print(f"Nilai Kuis        : {kuis}")
print(f"Nilai Praktikum   : {praktikum}")
print(f"Nilai UTS         : {uts}")
print(f"Nilai UAS         : {uas}")
print(f"Nilai Akhir       : {nilai_mahasiswa}")
print(f"Standar Kelulusan : {standar_kelulusan}")
print(f"Apakah Lulus?     : {apakah_lulus}")
print(f"Perlu Remedial?   : {perlu_remedial}")
```

1. Modifikasi input agar nilai belanja bisa dimasukkan melalui keyboard
2. Ubah diskon menjadi 35% jika belanja di atas 1.000.000


```python
#Kode Program Dari Materi
belanja = 850000
diskon = (0, 0.1 * belanja)[belanja > 500000]
total_harga = belanja - diskon
print("Total Harga Setelah Diskon =", total_harga)
```

```python
#Nilai Belanja
print("#Nilai Belanja")

#Memasukkan Data
belanja = float(input("Masukkan Total Belanja:"))

#Proses
diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon

#Output
print("\n#Hasil Perhitungan")
print("Total Belanja              =", belanja)
print("Diskon                     =", diskon)
print("Total Harga Setelah Diskon =", total_harga)
```
